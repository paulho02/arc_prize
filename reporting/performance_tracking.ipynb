```python
import matplotlib.pyplot as plt
import json
```

Due the the suspicion that the the time each iteration in the main loops increases the more iteration have been executed, this analysis aims to analyze memory usage over time:


```python

with open('../reporting/data_repo/tracking_obj_20250807_165302.json', 'r') as f:
    tracking_obj = json.load(f)
plt.plot(range(len(tracking_obj["memory_usage"])),
         tracking_obj["memory_usage"])
plt.xlabel('Iteration')
plt.ylabel('Memory Usage (MB)')
plt.title('Memory Usage Over Iterations')
plt.show()
```

with 500 iteration the increase proceeds:
